# 01 - Data Loading

This notebook verifies that CV, GCD, and EIS datasets are available and readable.

```python
from pathlib import Path

from src.config import CV_DATA_DIR, GCD_DATA_DIR, EIS_DATA_DIR
from src.io_utils import list_excel_files

print("CV directory:", CV_DATA_DIR)
print("GCD directory:", GCD_DATA_DIR)
print("EIS directory:", EIS_DATA_DIR)
```

```python
cv_files = list_excel_files(CV_DATA_DIR)
gcd_files = list_excel_files(GCD_DATA_DIR)
eis_files = list_excel_files(EIS_DATA_DIR)

print(f"CV files: {len(cv_files)}")
print(f"GCD files: {len(gcd_files)}")
print(f"EIS files: {len(eis_files)}")

for label, files in [("CV", cv_files), ("GCD", gcd_files), ("EIS", eis_files)]:
    print(f"\n{label} files:")
    for file in files:
        print(" -", file.name)
```
